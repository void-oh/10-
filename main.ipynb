## Step1 — 라이브러리 불러오기, 데이터 읽기

```python
# 데이터 읽기용 코드셀입니다.

import pandas as pd
print('='*20, 'student_dropout.csv 데이터 정보입니다.', '='*20)
tmp_df = pd.read_csv(r'C:\Users\user\OneDrive - stu.ac.kr\바탕 화면\DA\과제\10주차\Q9,Q10\drive-download-20261118T093929Z-1-001\student_dropout.csv')
display(tmp_df.shape)
display(tmp_df.columns)
display(tmp_df[['Age at enrollment', 'Admission grade', 'Scholarship holder', 'Target']].head())
```

## Step2 — 데이터 전처리

<b>코드 Cell</b>
- features 와 target 으로 데이터를 분리합니다.
- features 는 <b>크기만</b>, target 은 <b>크기와 앞부분</b>을 출력합니다.

```python
features = tmp_df.drop(columns=['Target'])
target = tmp_df['Target']
print('='*20,'features 정보입니다.' ,'='*20)
print(features.shape)
print('='*20,'target 정보입니다.' ,'='*20)
print(target.shape)
print(target.head())
```

<b>코드 Cell</b>
- features 와 target 데이터를 train_test_split 을 통해 분리합니다.
- test_size = 0.2 로 설정하고 random_state = 42 로 설정합니다.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=42)
print('='*20,'분할 이후 train 데이터 크기입니다.' ,'='*20)
print(X_train.shape)
print(y_train.shape)
print('='*20,'분할 이후 test 데이터 크기입니다.' ,'='*20)
print(X_test.shape)
print(y_test.shape)
print('='*20,'train 데이터와 test 데이터 비율 계산 결과입니다.' ,'='*20)
print(len(X_train)/len(features))
print(len(X_test)/len(features))
```

<b>코드 Cell</b>
- target 데이터에 대한 LabelEncoding 과정을 수행합니다.
- <b>이 셀에서는 출력이 발생하지 않습니다.</b>

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test)
```

<b>코드 Cell</b>
- hist 시각화를 통해 Label Encoding 전후의 모습을 확인합니다.

```python
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

fig, axes = plt.subplots(2, 1, figsize=(6, 4))

axes[0].hist(y_train)
axes[0].set_title('='*20 + ' LabelEncoding 이전의 분포입니다. ' + '='*20)

axes[1].hist(y_train_encoded)
axes[1].set_title('='*20 + ' LabelEncoding 이후의 분포입니다. ' + '='*20)

plt.tight_layout()
plt.show()
```

## Step3 — 모델 학습

<b>코드 Cell</b>
- 모델을 선언하고 학습시킨 뒤, 예측과 평가 지표 계산까지 수행합니다.
- <b>이 셀에서는 출력이 발생하지 않습니다.</b>

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train_encoded)

y_pred = model.predict(X_test)
acc = accuracy_score(y_test_encoded, y_pred)
report = classification_report(y_test_encoded, y_pred, target_names=le.classes_)
```

<b>코드 Cell</b>
- 앞 셀에서 계산한 정확도와 classification_report 를 출력합니다.

```python
print('='*20,'accuracy 결과입니다.' ,'='*20)
print(acc)
print('='*20,'classification_report 결과입니다.' ,'='*20)
print(report)
```

<b>코드 Cell</b>
- 학습한 모델이 중요하게 본 feature 를 정리하여 출력합니다.

```python
import pandas as pd

print('='*20, 'feature 중요도 상위 10개입니다.', '='*20)

fi_df = pd.DataFrame({
    'feature': features.columns,
    'importance': model.feature_importances_
})

fi_top10 = fi_df.sort_values(by='importance', ascending=False).head(10).reset_index(drop=True)
fi_top10['importance'] = fi_top10['importance'].round(4)

display(fi_top10)
```
